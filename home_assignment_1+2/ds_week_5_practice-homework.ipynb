```python
from sklearn.cluster import DBSCAN

dbs_eps = 0.5
dbs_min_sam = 5

dbscan = DBSCAN(
    eps = dbs_eps,
    min_samples=dbs_min_sam
).fit(X)
```

```python
from sklearn.metrics import silhouette_score

data['cluster'] = dbscan.labels_
labels = data['cluster'].to_numpy()
data
data.count(["cluster"] == -1)
data["cluster"].value_counts()
n_clusters = len(set(data['cluster']) - {-1})

n_noise = (labels == -1).sum()
print("Number of noise points:", n_noise)


print(f"Parameter values used: \n\teps = {dbs_eps}\n\tmin_samples = {dbs_min_sam}")
print("Number of clusters found (excluding noise):",n_clusters)
print("Number of noise points:", ) 

shl_score = silhouette_score(X, 
```
