```python
data = pd.read_csv("data/processed/Airbnb_Open_Data_Cleaned.csv", low_memory=False)

data
# cleaning the dataset, dropping null an unnecessary values

data = data.drop(["license","country", "country_code","house_rules","name", "host_name","neighbourhood","neighbourhood_group", "last_review","service_fee"], axis=1)

data["minimum_nights"] =  data["minimum_nights"] >= 1
data["availibility_365"] = (data["availability_365"] > 0) | (data["availability_365"] < 365)


data = data.dropna()
data.info()
data.isna().sum()
display(data)


output_path = Path("data/processed/Airbnb_Open_Data_dropped_uneccesarry_columns.csv")
data.to_csv(output_path, index=False)
```

```python
# Sorting categorical columns
categorical_columns = data.select_dtypes(include=["object", "category"]).columns.tolist()

data[categorical_columns].nunique()

data
```

```python
#Encoding categories
data.join(pd.get_dummies(categorical_columns))
display(data)



X = data.drop(["price"], axis=1)
y = data["price"]

```
