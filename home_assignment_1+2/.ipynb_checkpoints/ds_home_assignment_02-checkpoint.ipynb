```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.dpi"] = 110
```

```python
# load ingthe cleaned dataset from the first assignment
data_path = Path("data/processed/Airbnb_Open_Data_Cleaned.csv")

df = pd.read_csv(data_path, low_memory=False, parse_dates=["last_review"])

print("Dataset loaded:", df.shape)
display(df.head())
```

```python
# checking main columns 
numerical_columns = ["price", "minimum_nights"]

display(df[numerical_columns].describe().round(2).T)
```

```python
# 1. task : Run univariate_profile() or equivalent
def univariate_profile(dataframe, column):
    series = dataframe[column].dropna()

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    print("=" * 50)
    print(f"Univariate Profile: {column}")
    print("=" * 50)
    print(f"Count    : {len(series)}")
    print(f"Missing  : {dataframe[column].isnull().sum()}")
    print(f"Mean     : {series.mean():.3f}")
    print(f"Median   : {series.median():.3f}")
    print(f"Mode     : {series.mode()[0]}")
    print(f"Std Dev  : {series.std():.3f}")
    print(f"Min      : {series.min()}")
    print(f"Q1       : {q1}")
    print(f"Q3       : {q3}")
    print(f"Max      : {series.max()}")
    print(f"IQR      : {iqr}")
    print(f"Skewness : {series.skew():.3f}")

    fig, axes = plt.subplots(1, 4, figsize=(18, 4))
    fig.suptitle(f"Univariate Profile: {column}", fontsize=14)

    sns.histplot(series, kde=True, bins=15, ax=axes[0], color="steelblue")
    axes[0].axvline(series.mean(), color="red", linestyle="--", label="Mean")
    axes[0].axvline(series.median(), color="green", linestyle="--", label="Median")
    axes[0].set_title("Histogram + KDE")
    axes[0].legend(fontsize=8)

    sns.boxplot(y=series, ax=axes[1], color="lightsteelblue", width=0.4)
    axes[1].set_title("Boxplot")

    sns.violinplot(y=series, ax=axes[2], color="mediumpurple", inner="box")
    axes[2].set_title("Violin Plot")

    sns.ecdfplot(series, ax=axes[3], color="steelblue", linewidth=2)
    axes[3].set_title("Cumulative Distribution")

    plt.tight_layout()
    plt.show()
```

```python
# task 2 :Identify any skewness, outliers, or interesting patterns

# a profile of orughly symmetric column
univariate_profile(df, "price")

# a profile of a strongly skewed column
univariate_profile(df, "minimum_nights")
```

```python
#Find outliers in minimum_nights with the IQR method
q1 = df["minimum_nights"].quantile(0.25)
q3 = df["minimum_nights"].quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

minimum_night_outliers = df[
    (df["minimum_nights"] < lower_limit) |
    (df["minimum_nights"] > upper_limit)
]

print(f"Lower limit: {lower_limit}")
print(f"Upper limit: {upper_limit}")
print(f"Number of outliers: {len(minimum_night_outliers)}")
print(f"Outlier percentage: {len(minimum_night_outliers) / df['minimum_nights'].count() * 100:.2f}%")

display(minimum_night_outliers[["name", "room_type", "minimum_nights"]].head(10))
```

## Findings

The price distribution is almost symmetric, with a skewness close to zero and very similar mean and median values of about $625 and $624. 
The prices range from $50 to $1,200, but the IQR method does not identify any price outliers. In contrast, minimum_nights is extremely right-skewed, with a skewness of 86.70, while its mean of 8.14 nights is much higher than its median of 3 nights. Using the IQR rule, values below -2.5 or above 9.5 are clasified as outliers , resulting in 18,394 outlying observations. The minimum of -1,223 nights and maximum of 5,645 nights are not realistic and probably indicate data quality errors. The median and IQR describe minimum_nights better than the mean and standard deviation
